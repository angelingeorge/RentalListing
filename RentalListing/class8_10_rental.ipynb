```python
basic_features=["bathrooms","bedrooms","price","latitude","longitude","feature_count","photo_count","listing_hour","listing_dayofweek"]
x_basic=rental[basic_features].copy()
y=rental["interest_level"].copy()
print("x:",x_basic.shape)
print("y:",y.shape)
```

```python
pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x_basic,y,test_size=.20,random_state=42,stratify=y)
print("training:",x_train.shape)
print("testing:",x_test.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numeric_preprocessor=Pipeline([("imputer",SimpleImputer(strategy="median")),("scaler",StandardScaler())])
print("Numerical preprocessing pipeline created")
```

```python
from sklearn.linear_model import LogisticRegression
logistic_basic=Pipeline([("preprocessing",numeric_preprocessor),
                         ("model",LogisticRegression(max_iter=1000))])
logistic_basic.fit(x_train,y_train)
basic_pred=logistic_basic.predict(x_test)
print("Baseline logistic regression trained")
```

```python
from sklearn.metrics import(accuracy_score,classification_report,
confusion_matrix,ConfusionMatrixDisplay,f1_score)
print("Accuracy:",round(accuracy_score(y_test,basic_pred),3))
print("Macro F1:",round(f1_score(y_test,basic_pred,average="macro"),3))
print("\nClassification report:")
print(classification_report(y_test,basic_pred,zero_division=0))

ConfusionMatrixDisplay.from_predictions(
    y_test,basic_pred,labels=["low","medium","high"],
    display_labels=["low","medium","high"])
plt.title("Baseline logistic regression")
plt.show()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_demo=TfidfVectorizer(stop_words="english",max_features=20)

text_demo=rental["description"].fillna("").astype(str)
tfidf_matrix_demo=tfidf_demo.fit_transform(text_demo)
print("TF-IDF matrix shape:",tfidf_matrix_demo.shape)
print("Example vocabulary:")
print(tfidf_demo.get_feature_names_out()[:20])
```

```python
numeric_features=basic_features
text_features="description"
text_preprocessor=TfidfVectorizer(stop_words="english",max_features=1000,ngram_range=(1,2))

from sklearn.compose import ColumnTransformer
combined_preprocesssor=ColumnTransformer(transformers=[("num",numeric_preprocessor,numeric_features),
                                                      ("text",text_preprocessor,text_features)],
                                         remainder="drop")
print("Combined numerical+ TF-IDF preprocessor created")
```

```python
X_combined = rental[numeric_features + [text_features]].copy()
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_combined,y,test_size = 0.2,random_state = 42,
    stratify = y)
logistic_tf_idf = Pipeline([("preprocessing", combined_preprocesssor),
    ("model", LogisticRegression( max_iter = 1000,
        C = 1.0))
])
logistic_tf_idf.fit(X_train_c, y_train_c)
tf_idf_pred = logistic_tf_idf.predict(X_test_c)
print("TF-IDF + Logistic Regression trained")
print("Accuracy: ", round(accuracy_score(y_test_c, tf_idf_pred), 3))
print("Macro F1 Score: ", round(f1_score(y_test_c, tf_idf_pred, average = "macro"), 3))
print("\nClassification report:")
print(classification_report(y_test, tf_idf_pred, zero_division = 0))

```

```python
model_comparison = pd.DataFrame({
    "Model" : [
        "Baseline Logistic Regression",
        "TF-IDF + Logistic Regression"
    ],
    "Accuracy" : [
        accuracy_score(y_test, basic_pred),
        accuracy_score(y_test_c, tf_idf_pred)
    ],
    "Macro F1 Score" : [
        f1_score(y_test, basic_pred, average = "macro"),
        f1_score(y_test_c, tf_idf_pred, average = "macro")
    ]
})

display(model_comparison.sort_values(by = "Macro F1 Score", ascending = False))
```

```python
def safe_list_length(value):
    if isinstance(value,(list,tuple)):
        return len(value)
    return 0
def feature_to_text(value):
    if isinstance(value,list):
        return " ".join(str(item) for item in value)
    return ""
rental["feature_count"]=rental["features"].apply(safe_list_length)
rental["photo_count"]=rental["photos"].apply(safe_list_length)
rental["features_text"]=rental["features"].apply(feature_to_text)

display(
    rental[["features","features_text","feature_count","photo_count","photos"]].head())
```

```python
numeric_features=basic_features
text_features="features_text"
text_preprocessor=TfidfVectorizer(stop_words="english",max_features=1000,ngram_range=(1,2))

from sklearn.compose import ColumnTransformer
combined_preprocesssor=ColumnTransformer(transformers=[("num",numeric_preprocessor,numeric_features),
                                                      ("text",text_preprocessor,text_features)],
                                                       remainder="drop"
)
print("Combined numerical + TF-IDF preprocessor created.")
```

```python
X_combined = rental[numeric_features +[text_features]].copy()
X_train_c,X_test_c,y_train_c,y_test_c= train_test_split(
    X_combined,
    y,
    test_size=.20,
    random_state=42,
    stratify=y
)
logistic_tfidf = Pipeline([
    ("preprocessing",combined_preprocesssor),
    ("model",LogisticRegression(
        max_iter=1000,
        C=1.0
    ))
])
logistic_tfidf.fit(X_train_c,y_train_c)
tfidf_pred=logistic_tfidf.predict(X_test_c)
print("TF-IDF + Logistic Regression Trained.")

print("Accuracy:", round(accuracy_score(y_test_c, tfidf_pred),3))
print("Macro F1:", round(f1_score(y_test_c, tfidf_pred, average="macro"),3))

print("\nClassification report:")
print(classification_report(y_test_c, tfidf_pred, zero_division=0))
```
